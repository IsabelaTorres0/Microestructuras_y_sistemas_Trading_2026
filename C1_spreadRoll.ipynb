```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Importamos archivos 

data = pd.read_csv("btc_project_train.csv")
data.head(20)
```

```python
#Eliminamos los valores nulos
data = data.dropna()
data
```

```python
data[["Datetime", "Close"]]
```

```python
precio = data["Close"]
delta = precio.diff().dropna()
cov = delta.cov(delta.shift(1).dropna())
```

```python
#Calculamos nuestro window
```

```python

#Usamos la función rolling window para calcular el spread
window = 20
data['spread'] = data['Close'].rolling(window=window).cov(data['Close'].shift(1))
data[['Datetime', 'Close', 'spread']].head(20)

```
