## Pregunta 3

### Mátriz de correlación

```python
cols = ["sma_10", "sma_40", "sma_100", "ema_20", "bb_mid", "bb_pct",
        "atr_14", "rsi_14", "macd", "macd_signal", "macd_diff",
        "stoch_k", "stoch_d"]

corr = ind[cols].corr()
corr
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 8))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(range(len(cols)))
ax.set_yticks(range(len(cols)))
ax.set_xticklabels(cols, rotation=90)
ax.set_yticklabels(cols)

for i in range(len(cols)):
    for j in range(len(cols)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center", fontsize=7)

fig.colorbar(im, label="correlación")
plt.title("Matriz de correlación entre indicadores")
plt.tight_layout()
plt.show()
```

Que tan parecidos son nuestros indicadores unos con otros, en la siguiente mátriz de correlación podemos ver que el SMA y ema tiene una correlación del 100% mientras cambia con las bollinger el siguiente con una correlación similar es el atr con 44%. Sin embargo lo demás ,amtienen un poca coorrelación entre si, siendo e bb_pct el que mantiene mejor correlación con otros. 

Además yo separaría en dos grupos nuestras métriccas aquellas que SMA y EMA que son similares con mayor correlación. Mientras tanto, las estocasticas, borderline, la diferencia entre medias moviles(MACD) mantienen mejor correlación. 

## Pregunta 4
### ¿Alguno mira al futuro?

```python
import indicadores

def test_lookahead(df, func_name, t, **kwargs):
    """
    Compara el valor en la posición t calculado con TODA la muestra
    vs calculado solo con datos hasta t (inclusive).
    """
    func = getattr(indicadores, func_name)

    full_result = func(df, **kwargs)
    partial_result = func(df.iloc[:t+1], **kwargs)

    if isinstance(full_result, pd.DataFrame):
        full_val = full_result.iloc[t]
        partial_val = partial_result.iloc[-1]
    else:
        full_val = full_result.iloc[t]
        partial_val = partial_result.iloc[-1]

    return full_val, partial_val

# prueba varios indicadores en un mismo punto t
t = 5000  # elige un t bien adentro de la serie, con suficiente historia detrás

for name, kwargs in [
    ("sma", {"window": 10}),
    ("ema", {"window": 20}),
    ("rsi", {"window": 14}),
    ("macd", {"slow": 26, "fast": 12, "signal": 9}),
    ("bollinger", {"window": 20, "n_std": 2.5}),
    ("atr", {"window": 14}),
    ("stochastic", {"window": 14, "smooth": 3}),
]:
    full_val, partial_val = test_lookahead(df, name, t, **kwargs)
    diff = (full_val - partial_val)
    print(name)
    print("  full:   ", full_val.values if hasattr(full_val, "values") else full_val)
    print("  partial:", partial_val.values if hasattr(partial_val, "values") else partial_val)
    print("  diff:   ", diff.values if hasattr(diff, "values") else diff)
    print()
    
```

Ninguno de los indicadores tienen un look ahead, como no hay diferencia entre el full y parcial entre las métricas cada ventana, no afecta a los indicadores y su diferenci. Probé con 2 a 3 datos y en todos me daba una diferencia de 0 lo cual prueba que esto no es exclusivo a un único punto.
